# Escalado
Conjunto de datos: iris (incluido en sklearn)
Este conjunto de datos es compacto y perfecto para demostrar técnicas de escalado.

```python
# Importar librerías necesarias
from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler, StandardScaler
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Cargar el conjunto de datos
iris = load_iris(as_frame=True)
data = iris.data
```

```python
# Visualizar las estadísticas iniciales
print("Estadísticas iniciales:")
data.describe()
```

```python
# Aplicar MinMaxScaler
minmax_scaler = MinMaxScaler()
data_minmax = pd.DataFrame(minmax_scaler.fit_transform(data), columns=data.columns)
```

```python
# Aplicar StandardScaler
standard_scaler = StandardScaler()
data_standard = pd.DataFrame(standard_scaler.fit_transform(data), columns=data.columns)
```

### 1. ¿Qué diferencias encuentras entre las estadísticas originales y las del escalado MinMax?


```python
Las estadísticas originales muestran los valores en sus escalas iniciales, mientras que MinMaxScaler transforma los datos para que todas las características tengan valores entre 0 y 1. Esto modifica estadísticas como la media, la desviación estándar y los cuartiles, pero conserva la relación proporcional entre los valores. Su objetivo es facilitar el trabajo de ciertos algoritmos de Machine Learning al utilizar escalas comparables.
```

```python
# Estadísticas de los datos originales
print("Estadísticas originales:")
display(data.describe())

# Estadísticas después de aplicar MinMaxScaler
print("Estadísticas con MinMaxScaler:")
display(data_minmax.describe())
```

### 2. Crea una gráfica de la distribución de la variable 'sepal length (cm)' antes y después del escalado.

```python
# Crear una figura con dos gráficas
plt.figure(figsize=(12, 5))

# Gráfica 1: Datos originales
plt.subplot(1, 2, 1)
plt.hist(data['sepal length (cm)'], bins=15, edgecolor='black')
plt.title('Antes del escalado')
plt.xlabel('Longitud del sépalo (cm)')
plt.ylabel('Frecuencia')

# Gráfica 2: Datos escalados
plt.subplot(1, 2, 2)
plt.hist(data_minmax['sepal length (cm)'], bins=15, edgecolor='black')
plt.title('Después del escalado (MinMax)')
plt.xlabel('Longitud del sépalo escalada')
plt.ylabel('Frecuencia')

# Ajustar espacios y mostrar
plt.tight_layout()
plt.show()
```

### 3. Crea una gráfica de dispersión entre 'sepal length (cm)' y 'sepal width (cm)' con los datos escalados por StandardScaler.

```python
# Crear la figura
plt.figure(figsize=(8, 6))

# Crear gráfica de dispersión
plt.scatter(
    data_standard['sepal length (cm)'],
    data_standard['sepal width (cm)'],
    alpha=0.7
)

# Agregar títulos y etiquetas
plt.title('Relación entre longitud y ancho del sépalo (StandardScaler)')
plt.xlabel('Longitud del sépalo escalada')
plt.ylabel('Ancho del sépalo escalado')

# Mostrar gráfica
plt.grid(True)
plt.show()
```
